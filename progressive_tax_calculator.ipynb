```python
class SimpleIncomeTax:
    """A class for calculating progressive income tax (basic structure)."""

    def __init__(self) -> None:
        # Define tax brackets: [maximum income in this bracket, tax rate]
        # Note: None represents the uncapped highest bracket.
        self.tax_brackets = [
            (150000, 0.0),    # 0 - 150,000: Tax-exempt
            (300000, 0.05),   # 150,001 - 300,000: 5%
            (500000, 0.10),   # 300,001 - 500,000: 10%
            (None, 0.15)      # 500,001 and above: 15%
        ]

    def calculate_tax(self, net_income: float) -> float:
        """Calculate tax from net income."""
        if not isinstance(net_income, (int, float)) or net_income < 0:
            raise ValueError("Net income must be a non-negative number.")

        total_tax = 0.0
        previous_limit = 0.0

        for limit, rate in self.tax_brackets:
            # If income falls into this bracket (or it's the highest uncapped bracket)
            if limit is None or net_income <= limit:
                taxable_amount = net_income - previous_limit
                total_tax += taxable_amount * rate
                break
            else:
                # If income exceeds this bracket, calculate max tax for this bracket
                taxable_amount = limit - previous_limit
                total_tax += taxable_amount * rate
                previous_limit = limit

        return total_tax

# Example usage
calculator = SimpleIncomeTax()
income = 350000
print(f"Income: {income:,} THB | Tax to pay: {calculator.calculate_tax(income):,.2f} THB")
```

```python
def calculate_flat_tax(income: float, tax_rate: float = 0.10) -> float:
    """
    Calculate flat tax based on total income.

    Parameters:
    income (float): Total income.
    tax_rate (float): Tax rate (default is 10% or 0.10).

    Returns:
    float: Calculated tax amount.

    Raises:
    ValueError: If income is negative.
    """
    try:
        if income < 0:
            raise ValueError("Income cannot be negative.")

        tax_amount = income * tax_rate
        return tax_amount

    except TypeError:
        print("Error: Please enter numeric values only.")
        return 0.0

# Example usage
print(f"Tax to pay: {calculate_flat_tax(50000):.2f} THB")
```

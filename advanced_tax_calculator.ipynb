```python
from typing import List, Tuple, Optional

# 1. Custom Exception
class NegativeIncomeError(Exception):
    """Custom exception raised when income is negative."""
    pass

class TaxCalculatorPro:
    """
    Advanced tax calculator supporting customizable progressive tax brackets.
    """

    def __init__(self, brackets: List[Tuple[Optional[float], float]]) -> None:
        """
        Initialize with specific tax brackets.
        brackets: List of tuples containing (income limit, tax rate).
        """
        self.brackets = brackets

    def calculate(self, income: float) -> float:
        """Calculate tax based on the injected brackets."""
        if income < 0:
            raise NegativeIncomeError(f"Cannot calculate tax for negative income: {income}")

        tax = 0.0
        prev_limit = 0.0

        for limit, rate in self.brackets:
            if limit is None:  # Uncapped highest bracket
                tax += (income - prev_limit) * rate
                break

            if income > limit:
                tax += (limit - prev_limit) * rate
                prev_limit = limit
            else:
                tax += (income - prev_limit) * rate
                break

        return tax

# Professional usage: Separate the business rules (tax policy) from the logic engine.
tax_policy_2026 = [
    (150000.0, 0.0),
    (300000.0, 0.05),
    (500000.0, 0.10),
    (750000.0, 0.15),
    (1000000.0, 0.20),
    (None, 0.25)
]

# Initialize the calculator with the 2026 tax policy
pro_calculator = TaxCalculatorPro(tax_policy_2026)

try:
    my_income = 800000
    my_tax = pro_calculator.calculate(my_income)
    print(f"Income: {my_income:,.2f} THB | Tax to pay: {my_tax:,.2f} THB")
except NegativeIncomeError as e:
    print(f"Data Error: {e}")
```
